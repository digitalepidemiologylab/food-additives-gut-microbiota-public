```python
import pandas as pd

from additives_microbiota.constants import ADDITIVE_VARS, MIXTURES
```

### Load Data

```python
meta_additives = pd.read_csv("../data/metadata_files/meta_mixtures.csv", index_col=0)
```

### Calculate Z-scores for food features

```python
def calculate_zscore(df, column):
    """
    Calculate z-score for a given column in a dataframe.
    """
    return (df[column] - df[column].mean()) / df[column].std()
```

```python
meta_additives[f"HEI_zscore"] = calculate_zscore(meta_additives, "HEI")

for fg in ["vegetables", "fruits", "oils_nuts", "fast_food"]:
    meta_additives[f"{fg}_zscore"] = calculate_zscore(meta_additives, fg + "_fg_eaten")

#### For Additives
for additive in ADDITIVE_VARS + MIXTURES + ["total_additive_intake"]:
    meta_additives[f"{additive}_zscore"] = calculate_zscore(meta_additives, additive)

#################################################################
# Variables to subtract from
base_vars = ["HEI", "fruits", "vegetables", "oils_nuts"]
# Variables to subtract
subtract_vars = ADDITIVE_VARS + MIXTURES + ["fast_food", "total_additive_intake"]

# Create all combinations of differences
for base in base_vars:
    for sub in subtract_vars:
        col_name = f"{base}_minus_{sub}"
        meta_additives[col_name] = (
            meta_additives[f"{base}_zscore"] - meta_additives[f"{sub}_zscore"]
        )
```

### Save metadata with zscores

```python
meta_additives.to_csv("../data/metadata_files/meta_zscores.csv")

print(meta_additives.shape)
```
